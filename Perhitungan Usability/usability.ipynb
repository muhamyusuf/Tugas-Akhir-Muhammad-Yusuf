```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch

# Mock data from 5 respondents (each with scores for 2 dimensions)
data = {
    "Respondent": [f"R{i}" for i in range(1, 6)],
    "Pragmatic Quality": [1.8, 2.1, 2.3, 1.9, 2.0],
    "Hedonic Quality": [2.0, 2.4, 2.1, 1.8, 2.3]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# ✅ Print mean of each dimension correctly
mean_pragmatic = np.mean(df["Pragmatic Quality"])
mean_hedonic = np.mean(df["Hedonic Quality"])
print(f"Mean Pragmatic Quality: {mean_pragmatic:.2f}")
print(f"Mean Hedonic Quality: {mean_hedonic:.2f}")

# Calculate mean per dimension
means = [mean_pragmatic, mean_hedonic]
dimensions = list(df.columns[1:])

# Define benchmark ranges and colors
categories = ["Bad", "Below Average", "Above Average", "Good", "Excellent"]
bounds = [-1.0, 0.8, 1.0, 1.5, 1.8, 2.5]
colors = ["#e74c3c", "#f39c12", "#f1c40f", "#82e0aa", "#27ae60"]

# Create figure
fig, ax = plt.subplots(figsize=(9, 5))
x_pos = np.arange(len(dimensions))

# Draw stacked color zones for each dimension
bar_width = 0.5
for x in x_pos:
    for i in range(len(bounds) - 1):
        ax.bar(
            x,
            bounds[i + 1] - bounds[i],
            bottom=bounds[i],
            color=colors[i],
            edgecolor='none',
            width=bar_width,
            align='center',
        )

# Plot mean as black diamond marker for each dimension
ax.scatter(x_pos, means, color='black', s=100, zorder=3, label='Mean (All Respondents)')

# Add small horizontal lines to indicate mean range visually
for i, mean in enumerate(means):
    ax.hlines(mean, x_pos[i] - bar_width / 2, x_pos[i] + bar_width / 2, color='black', lw=2)
    # Optional: label value above mean
    ax.text(x_pos[i], mean + 0.05, f"{mean:.2f}", ha='center', va='bottom', fontsize=9, fontweight='bold')

# Custom chart setup
ax.set_xticks(x_pos)
ax.set_xticklabels(dimensions, rotation=25, ha='right')
ax.set_ylabel('UEQ-S Mean Score (-3 to +3)')
ax.set_ylim(-1, 2.5)
ax.set_xlim(-0.6, 1.6)
ax.set_title('UEQ-S Benchmark Mapping (Pragmatic vs Hedonic)')

# Improved legend
legend_elements = [Patch(facecolor=colors[i], edgecolor='black', label=categories[i]) for i in range(len(categories))]
legend_elements.append(Patch(facecolor='black', label='Mean (All Respondents)'))
ax.legend(handles=legend_elements, loc='upper left', fontsize=9, frameon=True, facecolor='white',
          edgecolor='black', title='UEQ-S Benchmark Levels', title_fontsize=10)

plt.grid(True, axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch

# Mock data dari 5 responden dengan skor rendah (kategori "Bad" / kualitas buruk)
data = {
    "Respondent": [f"R{i}" for i in range(1, 6)],
    "Pragmatic Quality": [-0.8, -0.5, 0.2, -0.3, 0.0],
    "Hedonic Quality": [-0.9, -0.6, -0.4, -0.2, 0.1]
}

# Convert ke DataFrame
df = pd.DataFrame(data)

# Hitung rata-rata tiap dimensi
mean_pragmatic = np.mean(df["Pragmatic Quality"])
mean_hedonic = np.mean(df["Hedonic Quality"])
print(f"Mean Pragmatic Quality: {mean_pragmatic:.2f}")
print(f"Mean Hedonic Quality: {mean_hedonic:.2f}")

# Siapkan data untuk plotting
means = [mean_pragmatic, mean_hedonic]
dimensions = list(df.columns[1:])

# Definisi benchmark dan warna UEQ
categories = ["Bad", "Below Average", "Above Average", "Good", "Excellent"]
bounds = [-1.0, 0.8, 1.0, 1.5, 1.8, 2.5]
colors = ["#e74c3c", "#f39c12", "#f1c40f", "#82e0aa", "#27ae60"]

# Buat figure
fig, ax = plt.subplots(figsize=(9, 5))
x_pos = np.arange(len(dimensions))
bar_width = 0.5

# Warna benchmark (setiap kolom berdiri sendiri)
for x in x_pos:
    for i in range(len(bounds) - 1):
        ax.bar(
            x,
            bounds[i + 1] - bounds[i],
            bottom=bounds[i],
            color=colors[i],
            edgecolor='none',
            width=bar_width,
            align='center',
        )

# Plot titik mean
ax.scatter(x_pos, means, color='black', s=100, zorder=3, label='Mean (All Respondents)')

# Garis horizontal kecil di setiap mean
for i, mean in enumerate(means):
    ax.hlines(mean, x_pos[i] - bar_width / 2, x_pos[i] + bar_width / 2, color='black', lw=2)
    ax.text(x_pos[i], mean + 0.05, f"{mean:.2f}", ha='center', va='bottom', fontsize=9, fontweight='bold')

# Konfigurasi sumbu
ax.set_xticks(x_pos)
ax.set_xticklabels(dimensions, rotation=25, ha='right')
ax.set_ylabel('UEQ-S Mean Score (-3 to +3)')
ax.set_ylim(-1, 2.5)
ax.set_xlim(-0.6, 1.6)
ax.set_title('UEQ-S Benchmark Mapping (Pragmatic, Hedonic)')

# Legenda rapi
legend_elements = [Patch(facecolor=colors[i], edgecolor='black', label=categories[i]) for i in range(len(categories))]
legend_elements.append(Patch(facecolor='black', label='Mean (All Respondents)'))
ax.legend(
    handles=legend_elements,
    loc='upper left',
    fontsize=9,
    frameon=True,
    facecolor='white',
    edgecolor='black',
    title='UEQ-S Benchmark Levels',
    title_fontsize=10
)

plt.grid(True, axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
